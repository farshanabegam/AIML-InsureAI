## Feature 1: Policy Q&A Chatbot

### Step 1: Create Insurance FAQ Knowledge Base

In this step, we create a small insurance FAQ dataset that will be used as the knowledge base for the Policy Q&A Chatbot. The chatbot will answer only insurance-related questions using this context.

```python
insurance_faq = """
Q1: What is health insurance?
A: Health insurance helps cover medical expenses such as hospitalization, surgeries, and treatments.

Q2: What is life insurance?
A: Life insurance provides financial support to the nominee after the policyholder's death.

Q3: What documents are required to file an insurance claim?
A: Claim form, identity proof, policy document, hospital bills, and supporting documents.

Q4: How long does claim settlement take?
A: Claim settlement usually takes 7–30 working days depending on document verification.

Q5: What is a premium?
A: A premium is the amount paid periodically to keep the insurance policy active.

Q6: What is a deductible?
A: A deductible is the amount the policyholder pays before the insurer covers the remaining claim.

Q7: Can I renew my insurance policy?
A: Yes, most insurance policies can be renewed before their expiry date.

Q8: What is cashless hospitalization?
A: Cashless hospitalization allows treatment without paying upfront at network hospitals.

Q9: What is covered under motor insurance?
A: Motor insurance covers accidental damages, theft, fire, and third-party liabilities depending on the policy.

Q10: Can an insurance claim be rejected?
A: Yes, claims may be rejected due to policy exclusions, incorrect information, or insufficient documentation.
"""

print("Insurance FAQ Created Successfully!")
```

### Step 2: Create the System Prompt

In this step, we define a system prompt that instructs the Gemini model to answer only insurance-related questions. The chatbot uses the Insurance FAQ as its knowledge base and politely refuses to answer questions outside the insurance domain.

```python
system_prompt = f"""
You are a professional Insurance Policy Assistant.

Your responsibilities:
- Answer ONLY insurance-related questions.
- Use the following FAQ as your knowledge base.
- If the user asks anything unrelated to insurance,
  politely reply:
  "I can only answer insurance-related questions."

Insurance FAQ:

{insurance_faq}
"""

print("✅ System Prompt Created Successfully!")
```

### Step 3: Build the Policy Q&A Chatbot Function

In this step, we create a chatbot function that combines the system prompt with the user's question and sends it to the Gemini model.

```python
def policy_chatbot(user_question):

    prompt = f"""
    {system_prompt}

    User Question:
    {user_question}
    """

    response = client.models.generate_content(
        model="gemini-flash-latest",
        contents=prompt
    )

    return response.text
```

### Step 4: Test the Policy Q&A Chatbot

Test the chatbot using insurance-related and non-insurance-related questions.

```python
questions = [
    "What is health insurance?",
    "What documents are required for an insurance claim?",
    "What is a premium?",
    "Who is the Prime Minister of India?"
]

for q in questions:
    print("=" * 60)
    print("Question:", q)
    print()
    print("Answer:")
    print(policy_chatbot(q))
    print()
```

## Feature 2: Claim Summarizer

### Step 1: Create the Claim Summarizer Prompt

In this step, we create a prompt that instructs the Gemini model to summarize a long insurance claim into a structured format containing the incident, date, damage, and estimated severity.

```python
def summarize_claim(claim_description):

    prompt = f"""
    You are a Senior Insurance Claims Officer.

    Read the following insurance claim description carefully.

    Summarize it into the following format:

    Incident:
    Date:
    Damage:
    Estimated Severity:

    Claim Description:

    {claim_description}
    """

    response = client.models.generate_content(
        model="gemini-flash-latest",
        contents=prompt
    )

    return response.text
```

### Step 2: Test the Claim Summarizer

Provide a sample insurance claim description and verify that the Gemini model generates a structured summary.

```python
claim = """
On 12 July 2026, my car was involved in a road accident while traveling on the highway.
Another vehicle hit the rear side of my car, causing damage to the rear bumper,
tail lights, and trunk door. No one was injured, but the vehicle required towing
to the nearest service center. The estimated repair cost is around $3,500.
"""

summary = summarize_claim(claim)

print(summary)
```

## Prompt Engineering Technique 1: Zero-shot Prompting

### Objective

Zero-shot prompting asks the model to perform a task without providing any examples. The model relies only on the given instruction.

```python
zero_shot_prompt = """
Summarize the following insurance claim.

Claim:
My car was damaged in a road accident on 15 June 2026.
The front bumper and headlights were broken.
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=zero_shot_prompt
)

print(response.text)
```

## Prompt Engineering Technique 2: Few-shot Prompting

### Objective

Few-shot prompting provides examples before asking the model to perform the task.

```python
few_shot_prompt = """
Example 1

Claim:
My bike was stolen yesterday.

Summary:
Incident: Bike Theft
Severity: High

Example 2

Claim:
Heavy rain damaged my house roof.

Summary:
Incident: Roof Damage
Severity: Moderate

Now summarize this claim.

Claim:
My car was hit by another vehicle while parked.
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=few_shot_prompt
)

print(response.text)
```

## Prompt Engineering Technique 3: Role/Persona Prompting

### Objective

In Role Prompting, the AI is instructed to act as a specific professional. Here, the Gemini model acts as a Senior Insurance Claims Officer and generates a professional claim status email.

```python
role_prompt = """
You are a Senior Insurance Claims Officer.

Write a professional email informing the customer that the insurance claim is currently under review.

Customer Name: John Smith
Claim Number: CLM20260045
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=role_prompt
)

print(response.text)
```

## Prompt Engineering Technique 4: Step-by-Step Prompting

### Objective

Step-by-step prompting guides the Gemini model to analyze the insurance claim systematically before generating the final summary. This improves the accuracy and structure of the response.

```python
step_by_step_prompt = """
You are a Senior Insurance Claims Officer.

Analyze the following insurance claim step by step.

Step 1: Identify the incident.
Step 2: Identify the date.
Step 3: Identify the damages.
Step 4: Estimate the severity.
Step 5: Provide the final structured summary.

Claim:

On 18 August 2026, the insured vehicle was involved in a collision with another car.
The rear bumper, left-side doors, and rear lights were damaged.
The estimated repair cost is approximately $4,800.
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=step_by_step_prompt
)

print(response.text)
```

## Prompt Engineering Technique 5: Structured Output Prompting (JSON)

### Objective

Structured Output Prompting instructs the Gemini model to generate the response in a predefined JSON format. This makes the output easy to process programmatically and improves consistency.

```python
json_prompt = """
You are a Senior Insurance Claims Officer.

Read the following insurance claim.

Return ONLY valid JSON.

Claim:

On 5 January 2026, a warehouse caught fire due to an electrical short circuit.
The roof, electrical wiring, and stored inventory were damaged.
Estimated loss is $75,000.

Return the result in this format:

{
  "Incident": "",
  "Date": "",
  "Damage": "",
  "Estimated_Severity": ""
}
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=json_prompt
)

print(response.text)
```

## Prompt Engineering Technique 6: Weak Prompt vs Engineered Prompt

### Objective

This comparison demonstrates how a well-engineered prompt produces more structured, detailed, and useful responses than a simple weak prompt.

```python
# Weak Prompt
weak_prompt = """
Summarize this insurance claim.

Claim:
The insured vehicle collided with another vehicle on 10 July 2026.
The front bumper and headlights were damaged.
"""

# Engineered Prompt
engineered_prompt = """
You are a Senior Insurance Claims Officer.

Read the insurance claim carefully.

Provide the response in the following format:

Incident:
Date:
Damage:
Estimated Severity:

Claim:
The insured vehicle collided with another vehicle on 10 July 2026.
The front bumper and headlights were damaged.
"""

# Weak Prompt Response
print("=" * 60)
print("WEAK PROMPT OUTPUT")
print("=" * 60)

weak_response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=weak_prompt
)

print(weak_response.text)

```

```python
import time

print("\n" + "=" * 60)
print("ENGINEERED PROMPT OUTPUT")
print("=" * 60)

for attempt in range(3):
    try:
        engineered_response = client.models.generate_content(
            model="gemini-flash-latest",
            contents=engineered_prompt
        )

        print(engineered_response.text)
        break

    except Exception as e:
        print(f"Attempt {attempt+1} failed... Retrying in 5 seconds")
        time.sleep(5)
```

### Comparison Between Weak Prompt and Engineered Prompt

The weak prompt generated only a short summary of the insurance claim without a specific structure. In contrast, the engineered prompt instructed the model to return information in a predefined format (Incident, Date, Damage, and Estimated Severity). As a result, the engineered prompt produced a more organized, detailed, and useful response, making it suitable for insurance claim analysis and documentation.
